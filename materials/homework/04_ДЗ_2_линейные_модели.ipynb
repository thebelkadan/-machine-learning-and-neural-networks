### Задание 1 [3 балла]

Напиши свой класс `LogisticRegression`, реализующий обучение и предсказание.

```python
class LogisticRegression:
    def __init__(self, lr=0.01, iterations=1000):
        self.lr = lr
        self.iterations = iterations
        self.w = None
        self.b = None
        self.losses = []

    def initialize_parameters(self, X):
        """
        Инициализирует веса и смещение нулями
        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
        На выходе:
            ничего не возвращает
        """
        # Напиши код здесь
        n_features = X.shape[1]
        self.w = np.zeros(n_features)
        self.b = 0

    def _sigmoid(self, z):
        """
        Вычисляет сигмоидную функцию активации: 1 / (1 + e^(-z))
        Принимает:
            z (ndarray): входные значения
        Возвращает:
            sigmoid (ndarray): значения после сигмоиды
        """
        z = np.clip(z, -500, 500)  # Ограничиваем значения, чтобы избежать переполнения
        # Напиши код здесь
        sigmoid = 1 / (1 + np.exp(-z))

        return sigmoid

    def compute_loss(self, X, y):
        """
        Вычисляет функцию потерь: бинарная кросс-энтропия
            L = -1/n * Σ(y*log(p) + (1-y)*log(1-p))
            Подсказка: используй epsilon, чтобы не получать ошибку при попытке рассчитать логарифм нуля
        Принимает:
            X (ndarray): массив признаков
            y (ndarray): вектор меток классов (0 или 1)
        Возвращает:
            loss (float): значение функции потерь
        """
        # Напиши код здесь
        n = X.shape[0]
        epsilon = 1e-15

        z = np.dot(X, self.w) + self.b
        probs = self._sigmoid(z)
        probs = np.clip(probs, epsilon, 1 - epsilon)
        loss = -1/n * np.sum(y * np.log(probs) + (1 - y) * np.log(1 - probs))

        return loss

    def step(self, X, y):
        """
        Один шаг градиентного спуска по всей выборке

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
            y (ndarray): метки классов (0 или 1)

        Возвращает:
            ничего не возвращает, обновляет веса и смещение
        """
        # Напиши код здесь
        n = X.shape[0]

        z = np.dot(X, self.w) + self.b
        probs = self._sigmoid(z)

        dw = 1/n * np.dot(X.T, (probs - y))
        db = 1/n * np.sum(probs - y)

        self.w = self.w - self.lr * dw
        self.b = self.b - self.lr * db

    def fit(self, X, y):
        """
        Обучает логистическую регрессию на тренировочных данных

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки
            y (ndarray): метки классов (0 или 1)

        Возвращает:
            ничего не возвращает, обучает модель
        """
        # Напиши код здесь
        self.initialize_parameters(X)

        for i in range(self.iterations):
            loss = self.compute_loss(X, y)
            self.losses.append(loss)
            self.step(X, y)

    def predict_proba(self, X):
        """
        Делает вероятностное предсказание

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки

        Возвращает:
            probs (ndarray): предсказанные вероятности классов
        """
        z = np.dot(X, self.w) + self.b
        probs = self._sigmoid(z)
        return np.vstack([1 - probs, probs]).T

    def predict(self, X):
        """
        Делает бинарное предсказание: класс 0 или 1

        Принимает:
            X (ndarray, shape: [n_samples, n_features]): обучающие признаки

        Возвращает:
            preds (ndarray): предсказанные метки классов
        """
        # Напиши код здесь
        probs = self.predict_proba(X)
        preds = (probs[:, 1] >= 0.5).astype(int)

        return preds
```

Заполни тело функции `apply_transformations()`. В функции уже реализована базовая версия: «Оригинал» — без преобразования и `log1p(y)` — лог-преобразование. Необходимо добавить ещё два преобразования:
- квадратный корень;
- Yeo — Johnson, более универсальное преобразование, похожее на Box — Cox, но умеет работать с отрицательными значениями и нулями, для его реализации используй `sklearn.preprocessing.PowerTransformer(method="yeo-johnson")`.

Функция должна возвращать словарь transforms:
- где ключ — название преобразования (строка);
- значение — кортеж из трёх элементов:
   - `y_transformed` — преобразованный `y_train`;
   - `inverse_transform_func` — функция, выполняющая обратное преобразование;
   - `transform_func` — функция, применяющая преобразование (она пригодится позже для `y_test`).

```python
from scipy.stats import boxcox
from sklearn.preprocessing import PowerTransformer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import cross_val_score, GridSearchCV


def apply_transformations(y_train):
    y_train = np.asarray(y_train)
    transforms = {
        "Оригинал": (y_train, lambda x: x, lambda x: x),
        "log1p(y)": (np.log1p(y_train), np.expm1, np.log1p)
    }
    # Напиши код здесь
    sqrt_y = np.sqrt(y_train)
    transforms["sqrt(y)"] = (sqrt_y, lambda x: x**2, np.sqrt)

    pt = PowerTransformer(method='yeo-johnson')
    y_1 = pt.fit_transform(y_train.reshape(-1, 1)).ravel()
    transforms["Yeo-Johnson"] = (y_1, pt.inverse_transform, pt.transform)
    return transforms
```

Входные параметры функции `evaluate_model_cv`:

`config = {

Внутри функции:

2. Обучи модель на `X_train` и ` y_train_t`.
3. Получи лучшую модель из `grid.best_estimator_`.
4. Сделай предсказания на `X_test`:
   - сначала в трансформированном масштабе;
   - затем примени `inverse_func`, чтобы вернуть предсказания в оригинальный масштаб.
5. Рассчитай метрики качества — MAE и $R^2$.
6. Выведи результаты.


```python
def evaluate_model_cv(transform_name, config, y_train_t, y_test_t, y_test_true, inverse_func):
  grid = GridSearchCV(estimator=config["estimator"],
                      param_grid=config["params"],
                      cv=5,
                      scoring='r2',
                      n_jobs=-1)

  grid.fit(X_train, y_train_t)
  best_model = grid.best_estimator_
  y_pred_t = best_model.predict(X_test)
  y_pred = inverse_func(y_pred_t.reshape(-1, 1)).ravel()
  mae = mean_absolute_error(y_test_true, y_pred)
  r2 = r2_score(y_test_true, y_pred)

  print(f"{transform_name:15} | MAE: {mae:.2f} | R²: {r2:.4f}")
  return mae, r2

```
